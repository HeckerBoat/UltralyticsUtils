```python
import matplotlib.pyplot as plt
import numpy as np


def generate_pie_chart(data):
    labels = list(data.keys())
    percentages = list(data.values())

    plt.rcParams['font.sans-serif'] = ['Times New Roman']
    plt.rcParams['axes.unicode_minus'] = False
    plt.figure(figsize=(8, 6))
    plt.pie(percentages, labels=labels, autopct=lambda p: '{:.1f}%'.format(p), startangle=140,
            textprops={'fontsize': 12, 'fontstyle': 'italic', 'fontweight': 'bold'})
    plt.axis('equal')
    plt.savefig('../results/labels.png', dpi=300)
    plt.show()

data = {
    'Galinsoga parviflora': 19.4,
    'Alternanthera philoxeroides': 16.8,
    'Cerastium glomeratum': 15.9,
    'Cardamine hirsuta': 16.4,
    'Amaranthus retroflexus': 14.9,
    'Pilea peperomioides': 16.6,
}

generate_pie_chart(data)

```
